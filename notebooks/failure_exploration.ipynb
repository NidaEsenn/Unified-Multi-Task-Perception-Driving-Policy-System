# Failure Case Exploration

Deep dive into system failure modes with visualizations and analysis.

```python
import json
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
print("✓ Libraries loaded")
```

```python
# Load failure data
try:
    with open("../results/failures.json", 'r') as f:
        failures = json.load(f)
    print(f"✓ Loaded failure data for {failures.get('total_samples', 0)} samples")
    print(f"  Failed samples: {failures.get('failed_samples', 0)} ({failures.get('failure_rate', 0):.1f}%)")
except FileNotFoundError:
    print("⚠️ Run 'python scripts/analyze_failures.py' first")
    failures = {}
```

## Failure Mode Distribution

```python
if 'failure_modes' in failures:
    modes = failures['failure_modes']
    
    # Create detailed analysis
    for mode_name, mode_data in list(modes.items())[:5]:
        print(f"\n{'='*60}")
        print(f"Failure Mode: {mode_name}")
        print(f"{'='*60}")
        print(f"\nDescription: {mode_data.get('description', 'N/A')}")
        print(f"Affected: {mode_data.get('affected_samples', 0)} samples ({mode_data.get('percentage', 0):.1f}%)")
        
        print("\nRoot Causes:")
        for cause in mode_data.get('root_causes', []):
            print(f"  • {cause}")
        
        print("\nMitigation Strategies:")
        for strategy in mode_data.get('mitigation_strategies', []):
            print(f"  • {strategy}")
```
